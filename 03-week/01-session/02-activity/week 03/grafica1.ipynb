```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

# Subir archivo desde tu PC (te pedirá seleccionarlo)
uploaded = files.upload()

# Luego lo cargas en un DataFrame
df = pd.read_csv("us-software-engineer-jobs-zenrows.csv")  # ajusta el nombre exacto

# Ver primeras filas
df.head()

# Ver información general
df.info()



# ===============================
# 1. TOP 10 empresas con más ofertas
# ===============================
top_companies = df['company'].value_counts().head(10)

plt.figure(figsize=(10,6))
top_companies.plot(kind='bar', color='skyblue')
plt.title("Top 10 empresas con más ofertas de trabajo")
plt.xlabel("Empresa")
plt.ylabel("Cantidad de ofertas")
plt.xticks(rotation=45, ha='right')
plt.show()


# ===============================
# 2. Tipos de empleo (full-time, part-time, etc.)
# ===============================
job_types = df['types'].value_counts().head(5)

plt.figure(figsize=(6,6))
job_types.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
plt.title("Distribución de tipos de empleo")
plt.ylabel("")
plt.show()


# ===============================
# 3. Modalidad de trabajo (remoto, híbrido, presencial)
# ===============================
remote_models = df['remote_work_model'].value_counts()

plt.figure(figsize=(6,6))
remote_models.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=plt.cm.Set2.colors)
plt.title("Modelos de trabajo remoto")
plt.ylabel("")
plt.show()


# ===============================
# 4. Urgentemente contratando (Sí vs No)
# ===============================
urgent = df['urgently_hiring'].value_counts()

plt.figure(figsize=(6,6))
urgent.plot(kind='bar', color=['tomato','lightgreen'])
plt.title("¿Están contratando urgentemente?")
plt.xlabel("Estado")
plt.ylabel("Cantidad de ofertas")
plt.xticks(rotation=0)
plt.show()


# ===============================
# 5. Promedio de rating de empresas (histograma)
# ===============================
plt.figure(figsize=(8,5))
df['rating'].hist(bins=30, color='purple', alpha=0.7)
plt.title("Distribución de ratings de empresas")
plt.xlabel("Rating")
plt.ylabel("Frecuencia")
plt.show()
```

```python
from google.colab import files
import pandas as pd

# Subir archivo desde tu PC (te pedirá seleccionarlo)
uploaded = files.upload()

# Luego lo cargas en un DataFrame
df = pd.read_csv("Covid Data.csv")  # ajusta el nombre exacto

# Ver primeras filas
df.head()

# Ver información general
df.info()
import matplotlib.pyplot as plt

# ===============================
# 1. Distribución por sexo (0 = mujer?, 1 = hombre?, depende del diccionario de datos)
# ===============================
sex_counts = df['SEX'].value_counts()

plt.figure(figsize=(6,6))
sex_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['skyblue','lightcoral'])
plt.title("Distribución por sexo")
plt.ylabel("")
plt.show()


# ===============================
# 2. Tipo de paciente (1 = ambulatorio, 2 = hospitalizado)
# ===============================
patient_type = df['PATIENT_TYPE'].value_counts()

plt.figure(figsize=(6,6))
patient_type.plot(kind='bar', color=['green','red'])
plt.title("Tipo de paciente (Ambulatorio vs Hospitalizado)")
plt.xlabel("Tipo de paciente")
plt.ylabel("Cantidad de casos")
plt.xticks(rotation=0)
plt.show()


# ===============================
# 3. Intubados (0 = no, 1 = sí)
# ===============================
intubed = df['INTUBED'].value_counts()

plt.figure(figsize=(6,6))
intubed.plot(kind='bar', color=['orange','blue'])
plt.title("Pacientes intubados")
plt.xlabel("Estado")
plt.ylabel("Cantidad de casos")
plt.xticks(rotation=0)
plt.show()


# ===============================
# 4. Distribución de edades
# ===============================
plt.figure(figsize=(10,6))
df['AGE'].hist(bins=50, color='purple', alpha=0.7)
plt.title("Distribución de edades de pacientes")
plt.xlabel("Edad")
plt.ylabel("Cantidad de casos")
plt.show()


# ===============================
# 5. Comorbilidades principales
# ===============================
comorbidities = ['DIABETES','COPD','ASTHMA','HIPERTENSION','OBESITY','RENAL_CHRONIC','TOBACCO']
comorb_counts = df[comorbidities].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
comorb_counts.plot(kind='bar', color='teal')
plt.title("Comorbilidades más frecuentes en pacientes COVID")
plt.xlabel("Condición")
plt.ylabel("Cantidad de pacientes con la condición")
plt.xticks(rotation=45)
plt.show()


# ===============================
# 6. Estado en UCI (ICU)
# ===============================
icu = df['ICU'].value_counts()

plt.figure(figsize=(6,6))
icu.plot(kind='bar', color=['lightgreen','red'])
plt.title("Pacientes en UCI")
plt.xlabel("Estado")
plt.ylabel("Cantidad de casos")
plt.xticks(rotation=0)
plt.show()
```
